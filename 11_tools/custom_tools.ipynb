```python
from langchain.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int) -> int:
    """Add two numbers and return result"""
    return a + b
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
```

```python
llm_with_tool = llm.bind_tools([add_numbers])
```

```python
response = llm_with_tool.invoke("What is 25 + 75 ?")
```

```python
25 + 75
```

```python
from langchain.messages import HumanMessage

messages = [
    HumanMessage(
        content="What is 25 + 75?"
    )
]

response = llm_with_tool.invoke(messages)

print(response.tool_calls)
```

```python
#add_numbers.invoke(response.tool_call.args)
```

```python
for tool_call in response.tool_calls:

    result = add_numbers.invoke(tool_call)

    print(result)
```

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers],
    system_prompt="You are a helpful mathematical assistant."
)
```

```python
result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "What is 125 + 375?"
        }
    ]
})
```
